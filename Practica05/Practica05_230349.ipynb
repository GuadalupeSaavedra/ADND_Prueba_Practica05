```python
import plotly.graph_objs as go
from plotly import tools
from plotly.offline import iplot

def bar_hor(df, col, limit=100, color='#b0c4de', rev=False):
    # DIAGNÓSTICO: Esto imprimirá en tu celda qué datos está leyendo realmente
    print(f"--- Datos leídos para la columna '{col}' (rev={rev}) ---")
    cnt_srs = df[col].value_counts()
    print(cnt_srs.head(limit)) # Veremos en texto si son 3 o 10 estados
    
    if rev:
        srs = cnt_srs.tail(limit)
        yy = [str(index) for index in srs.index]
        xx = [float(val) for val in srs.values]
    else:
        srs = cnt_srs.head(limit)
        yy = [str(index) for index in srs.index][::-1]
        xx = [float(val) for val in srs.values][::-1]
        
    # Forzamos explícitamente a Plotly a tratar el eje X como numérico y el Y como categoría
    trace = go.Bar(
        y=yy, 
        x=xx, 
        orientation='h', 
        marker=dict(color=color),
        type='bar'
    )
    return trace

# --- Configuración y despliegue ---

# Asegúrate de que 'Donor State' sea la columna correcta. 
# Si tu dataset tiene las siglas en otra columna (como 'State'), cámbialo aquí:
trace2 = bar_hor(donors_df, 'Donor State', limit=10, color='#c2d2ed', rev=False)
trace4 = bar_hor(donors_df, 'Donor State', limit=10, color='#c2d2ed', rev=True)

fig = tools.make_subplots(
    rows=1, 
    cols=2, 
    print_grid=False, 
    subplot_titles=['States with maximum Donors', 'States with minimum Donors']
)

fig.append_trace(trace2, 1, 1)
fig.append_trace(trace4, 1, 2)

# Forzamos los ejes en el layout para que no se queden en (0, 1)
fig['layout'].update(
    height=450, 
    width=900,
    showlegend=False,
    margin=dict(l=100, r=40, t=80, b=40),
    plot_bgcolor='rgba(0,0,0,0)',
    xaxis1=dict(autorange=True, type='linear'),
    xaxis2=dict(autorange=True, type='linear'),
    yaxis1=dict(type='category'),
    yaxis2=dict(type='category')
)

iplot(fig)
```

## 2.6 Cantidad de donantes docentes y no docentes por estado

```python
# which city and state has maximum teacher donors 
tempdf = donors_df.groupby(['Donor State', 'Donor Is Teacher']).agg({'Donor Is Teacher' : 'count' }).rename(columns={'Donor Is Teacher' : 'Teacher Donor Counts'}).reset_index()

no_df = tempdf[tempdf['Donor Is Teacher']=='No']
no_x = no_df['Donor State']
no_y = no_df['Teacher Donor Counts']
yes_df = tempdf[tempdf['Donor Is Teacher']=='Yes']
yes_x = yes_df['Donor State']
yes_y = yes_df['Teacher Donor Counts']

trace1 = go.Bar(
    x=no_x,
    y=no_y,
    name='Non Teacher Donors',
    marker=dict(color='#f259d6'),
    opacity=0.8
)
trace2 = go.Bar(
    x=yes_x,
    y=yes_y,
    name='Teacher Donors',
    marker=dict(color='#f7bb31'),
    opacity=0.8
)

data = [trace1, trace2]
layout = go.Layout(
    barmode='group',
    legend=dict(dict(x=-.1, y=1.2)),
    margin=dict(b=120),
    title = 'States and the distribution of Teacher and Non Teacher Donors',
)

fig = go.Figure(data=data, layout=layout)
iplot(fig, filename='grouped-bar')
```
